```python
#Given a list of integers and a number k, find the maximum sum of any k consecutive elements.

def max_sum(numbers,k):
  n = len(numbers)
  window_sum =sum(numbers[0:k])
  max_sum = window_sum
  for i in range(n-k):
    window_sum = window_sum -numbers[i]+numbers[i+k]
    if window_sum > max_sum:
      max_sum = window_sum
  return max_sum




numbers = [2, 1, 5, 1, 3, 2]
k = 3
print(max_sum(numbers,k))
```

```python
#Day 28 Given an array of integers, determine whether it is a palindrome.

"""
Time Complexity = O(n)
Space Complexity = O(1)
"""
def find_palindrome(numbers):
    left = 0
    right = len(numbers) - 1

    while left < right:
        if numbers[left] != numbers[right]:
            return False

        left += 1
        right -= 1

    return True

numbers = [1, 2, 3, 3, 1]
print(find_palindrome(numbers))
```

```python
#Day 27 Given two sorted arrays, merge them into a single sorted array.
"""
Time Complexity = O(n)
Space Complexity = O(n)
"""

def merge_array(nums1,nums2):
  i,j =0,0
  merged=[]
  while(i<len(nums1)) and (j<len(nums2)):
    if nums1[i]<=nums2[j]:
      merged.append(nums1[i])
      i+=1
    else:
      merged.append(nums2[j])
      j+=1
  merged.extend(nums1[i:])
  merged.extend(nums2[j:])
  return merged



nums1 = [1, 3, 8]
nums2 = [2, 4,5,6,9]
print(merge_array(nums1,nums2))
```

```python
#26 Given a sorted list of integers, remove all occurrences of a target value and return the remaining elements. 2 pointer
"""
Time Complexity = O(n)
Space Complexity = O(1)
"""

def remove_occurence(numbers,target):
  wp =0
  for rp in range(len(numbers)):
    if numbers[rp]!= target:
      numbers[wp]=numbers[rp]
      wp+=1
  return numbers[:wp]




numbers = [1, 2, 3, 3, 4, 5]
target = 3
print(remove_occurence(numbers,target))

```

```python
#Day 25 Given a string, reverse it using the Two Pointer technique.
"""
Time Complexity = O(n)
Space Complexity = O(n) - which creates a new list of size n.
"""

def reverse_string(text):
  text = list(text)
  l=0
  r=len(text)-1
  while(l<r):
    text[l],text[r]=text[r],text[l]
    l+=1
    r-=1
  return "".join(text)


text = "nandu"
print(reverse_string(text))
```

```python
#Day 21 Given a list of integers, find the maximum sum of any two numbers. comeback

"""
Time Complexity = O(n)
Space Complexity = O(1)
"""

def maximum_sum(numbers):
  largest = float("-inf")
  second = float("-inf" )
  for n in numbers:
    if n> largest:
      second = largest
      largest = n
    elif n >second:
      second = n
  return largest+second


numbers = [1, 5, 3, 9]
print(maximum_sum(numbers))
```

```python
#Day 20 Given a list of integers and a target value, return True if any two numbers add up to the target, otherwise return False.

"""
Time Complexity = O(n)
Space Complexity = O(n)
"""
def find_target(numbers, target):
    seen = set()

    for n in numbers:
        need = target - n

        if need in seen:
            return True

        seen.add(n)

    return False


numbers = [3,2]
target = 6
print(find_target(numbers,target))
```

```python
#Day 18 Given a list of integers, find the intersection of two lists.
"""
Time Complexity = O(n) - n number
Space Complexity = O(n) - n number
"""
def find_intersection(list1,list2):
  seen = set()
  list2 = set(list2)
  for n in list1:
    if n in list2:
      seen.add(n)
  return list(seen)

nums1 = [1, 2, 2, 1]
nums2 = [3, 4, 5, 6]
print(find_intersection(nums1,nums2))

```

```python
#Day 14 Given a list of integers containing numbers from 1 to n, with exactly one number missing, find the missing number.
"""
Time Complexity = O(n) loop runs n num times
Space Complexity = O(1) normal spacing
"""

def find_missing_number(numbers):
  value = numbers[0]
  for n in numbers:
    if n == value:
      value +=1
    else:
     return value

numbers = [1, 2, 3, 5]
print(find_missing_number(numbers))
```

```python
#Day 7 Given a list of integers, return the second largest number.

def secondLargest(numbers):
  for i in range(len(numbers)):
    for j in range(len(numbers)-1):
      if numbers[j]>numbers[j+1]:
        temp = numbers[j]
        numbers[j] = numbers[j+1]
        numbers[j+1] = temp
  return numbers[-2]


numbers = [5, 12, 3, 20, 15]
print(secondLargest(numbers))
```
